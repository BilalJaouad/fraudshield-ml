```python
def replace_negative_by_NaN(train,validate,test,cols_missing,cols_scaling) :

  train = train.copy()
  validate = validate.copy()
  test = test.copy()

  for df in [train,validate,test] :
    for col in cols_missing :

      df.loc[df[col] ==-1, col] = np.nan

      df[f"missing_{col}"] = df[col].isna().astype(int)
  imputer = SimpleImputer(strategy="median")
  imputer.fit(train[cols_missing])

  for df in [train, validate, test]:
    df[cols_missing] = imputer.transform(df[cols_missing])

  scaler = StandardScaler()
  scaler.fit(train[cols_scaling])
  for df in [train, validate, test]:
    df[cols_scaling] = scaler.transform(df[cols_scaling])
  return train, validate, test
```

```python

def prepa_balcon_amount(X) :
  X.loc[X["intended_balcon_amount"]<0,"intended_balcon_amount"] = np.nan
  X["missing_intended_balcon_amount"] = X["intended_balcon_amount"].isna().astype(int)
```

```python
def imputing_scaling(X_train,X_val,X_test) :
  X_train = X_train.copy()
  X_val = X_val.copy()
  X_test = X_test.copy()
  for df in [X_train,X_val,X_test] :
    prepa_balcon_amount(df)
  imputer = SimpleImputer(strategy="median")
  scaler = StandardScaler()
  imputer.fit(X_train[["intended_balcon_amount"]])

  for df in [X_train,X_val,X_test] :
    df["intended_balcon_amount"] = imputer.transform(df[["intended_balcon_amount"]]).ravel()
  scaler.fit(X_train[["intended_balcon_amount"]])
  for df in [X_train,X_val,X_test] :

    df["intended_balcon_amount"] = scaler.transform(df[["intended_balcon_amount"]])
  return X_train , X_val,X_test
```

```python
def optimizing(model,X_validat,y_val,avg_preci_score=False) :
  y_proba = model.predict_proba(X_validat)[:, 1]
  precision, recall, thresholds = precision_recall_curve(y_val, y_proba)
  f1_scores = 2 * precision[:-1] * recall[:-1] / (
    precision[:-1] + recall[:-1] + 1e-10
)
  best_idx = np.argmax(f1_scores)
  best_threshold = thresholds[best_idx]

  y_pred = (y_proba >= best_threshold).astype(int)
  if avg_preci_score :
    ap = average_precision_score(y_val, y_proba)
    return best_threshold,y_pred,precision,recall,thresholds,f1_scores,best_idx,y_proba,ap
  return best_threshold,y_pred,precision,recall,thresholds,f1_scores,best_idx,y_proba





```

```python
best_threshold,y_pred,precision,recall,thresholds,f1_scores,best_idx,y_proba,ap = optimizing(model1,X_validat,y_val,avg_preci_score=True)
print(classification_report(y_val, y_pred, zero_division=0))
print("Best threshold:", best_threshold)
print("Best F1-score:", f1_scores[best_idx])
print("average precision score : ",ap)
```

```python
C_fp = 100
C_fn = 1000

y_true = np.asarray(y_compare)
y_proba = np.asarray(y_proba_initial)

def evaluate_policy(name, y_pred):
    tn, fp, fn, tp = confusion_matrix(
        y_true, y_pred, labels=[0, 1]
    ).ravel()

    precision = tp / (tp + fp) if tp + fp > 0 else 0
    recall = tp / (tp + fn) if tp + fn > 0 else 0

    f1 = (
        2 * precision * recall / (precision + recall)
        if precision + recall > 0 else 0
    )

    prevalence = y_true.mean()
    lift = precision / prevalence if prevalence > 0 else 0

    return {
        "Policy": name,
        "Alerts": int(tp + fp),
        "TP": int(tp),
        "FP": int(fp),
        "FN": int(fn),
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "Alert Rate": y_pred.mean(),
        "Lift": lift,
        "Total Cost (MAD)": fp * C_fp + fn * C_fn
    }
```

```python

results = []


threshold_empirical = float(best_threshold)

results.append(
    evaluate_policy(
        "Optimal Cost Threshold",
        (y_proba >= threshold_empirical).astype(int)
    )
)


threshold_theoretical = C_fp / (C_fp + C_fn)

results.append(
    evaluate_policy(
        "Theoretical Threshold",
        (y_proba >= threshold_theoretical).astype(int)
    )
)


n = len(y_proba)
sorted_indices = np.argsort(-y_proba, kind="stable")

for pct in [0.01, 0.03, 0.05]:
    k = int(n * pct)

    y_pred = np.zeros(n, dtype=int)
    y_pred[sorted_indices[:k]] = 1

    results.append(
        evaluate_policy(f"Top {pct:.0%}", y_pred)
    )


rng = np.random.default_rng(42)

for pct in [0.01, 0.03, 0.05]:
    k = int(n * pct)

    selected = rng.choice(n, size=k, replace=False)

    y_pred = np.zeros(n, dtype=int)
    y_pred[selected] = 1

    results.append(
        evaluate_policy(f"Random {pct:.0%}", y_pred)
    )


results.append(
    evaluate_policy("No Detection", np.zeros(n, dtype=int))
)
```

```python

comparison_df = pd.DataFrame(results)

display(
    comparison_df
    .sort_values("Total Cost (MAD)")
    .style.format({
        "Precision": "{:.2%}",
        "Recall": "{:.2%}",
        "F1": "{:.3f}",
        "Alert Rate": "{:.2%}",
        "Lift": "{:.2f}x",
        "Total Cost (MAD)": "{:,.0f}"
    })
)
```

```python
# Test de bout en bout : observation BRUTE → transformateurs exportés → prédiction.
# N'ajuste AUCUN objet ici. Même ordre de transformations que l'entraînement.
def preprocess_raw_v2(raw_df):
    data = raw_df.copy()
    required = schema["raw_features"]
    missing = sorted(set(required) - set(data.columns))
    if missing:
        raise ValueError(f"Colonnes brutes manquantes : {missing}")
    data = data.loc[:, required].copy()

    for col in schema["cols_missing"]:
        data.loc[data[col] == -1, col] = np.nan
        data[f"missing_{col}"] = data[col].isna().astype(int)
    data[schema["cols_missing"]] = imputer_missing_loaded.transform(data[schema["cols_missing"]])
    data[schema["cols_scaling"]] = scaler_numeric_loaded.transform(data[schema["cols_scaling"]])

    col = "intended_balcon_amount"
    data.loc[data[col] < 0, col] = np.nan
    data[f"missing_{col}"] = data[col].isna().astype(int)
    data[col] = imputer_balcon_loaded.transform(data[[col]]).ravel()
    data[col] = scaler_balcon_loaded.transform(data[[col]]).ravel()

    assert data.columns.tolist() == schema["pre_encoding_features"], "Colonnes avant OHE incorrectes"
    cats = schema["categorical_columns"]
    encoded = pd.DataFrame(encoder_loaded.transform(data[cats]),
                           columns=encoder_loaded.get_feature_names_out(), index=data.index)
    final = pd.concat([data.drop(columns=cats), encoded], axis=1)
    assert final.columns.tolist() == schema["model_features"], "Ordre des features finales incorrect"
    return final

# Reproduire les données du mois 6 avant preprocessing, mêmes indices.
raw_reference = raw_X_v2.loc[X_reference.index].copy()
reconstructed = preprocess_raw_v2(raw_reference)
np.testing.assert_allclose(reconstructed.to_numpy(dtype=float),
                           X_reference.to_numpy(dtype=float), rtol=1e-7, atol=1e-9)
probabilities = model.predict_proba(reconstructed)[:, 1]
np.testing.assert_allclose(probabilities, reference_proba, rtol=1e-7, atol=1e-9)
print("✅ END-TO-END validé : brut → preprocessing exporté → CatBoost → même score")
```
